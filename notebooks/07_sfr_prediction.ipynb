# 07 — SFR Prediction

Pipeline smoke test only: verify deterministic train/test separation and a minimal model fit on synthetic fixtures. No observational performance is inferred.


```python3
import numpy as np
from sklearn.ensemble import RandomForestRegressor
rng=np.random.default_rng(42)
X=rng.normal(size=(20,3)); y=0.5*X[:,0]-0.2*X[:,1]+0.1*X[:,2]
X_train,X_test=X[:15],X[15:]; y_train,y_test=y[:15],y[15:]
model=RandomForestRegressor(n_estimators=16,random_state=42,n_jobs=1).fit(X_train,y_train)
pred=model.predict(X_test)
assert pred.shape==y_test.shape and np.isfinite(pred).all()
print('SFR model pipeline smoke test passed')
```
